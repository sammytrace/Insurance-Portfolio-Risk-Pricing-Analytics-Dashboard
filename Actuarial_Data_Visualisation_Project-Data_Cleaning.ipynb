### Derived Features

#### Risk_Tier

```python
def risk_tier(row):
    freq = row["Claims_Frequency"]
    sev = row["Claims_Severity"]

    if freq >= 3 or sev == "High":
        return "High Risk"
    elif freq == 2 or sev == "Medium":
        return "Medium Risk"
    else:
        return "Low Risk"

synthetic_insurance_data["Risk_Tier"] = synthetic_insurance_data.apply(risk_tier, axis=1)
```

#### Loyalty_Band

```python
def loyalty_band(prior_insurance):
    if pd.isna(prior_insurance):
        return "Unknown"
    elif prior_insurance == "<1 year":
        return "New Customer"
    elif prior_insurance == "1-5 years":
        return "1–5 Years"
    elif prior_insurance == ">5 years":
        return ">5 Years"
    else:
        return "Other"

synthetic_insurance_data["Loyalty_Band"] = synthetic_insurance_data["Prior_Insurance"].apply(loyalty_band)
```

#### Engagement Level

```python
# Weighted engagement score

synthetic_insurance_data["Engagement_Score"] = (
    synthetic_insurance_data["Website_Visits"] +
    2 * synthetic_insurance_data["Inquiries"] +
    3 * synthetic_insurance_data["Quotes_Requested"]
)

# Classify into bands
def engagement_level(score):
    if score <= 5:
        return "Low Engagement"
    elif score <= 15:
        return "Medium Engagement"
    else:
        return "High Engagement"

synthetic_insurance_data["Engagement_Level"] = synthetic_insurance_data["Engagement_Score"].apply(engagement_level)
```
